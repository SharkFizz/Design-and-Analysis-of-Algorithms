观察下方的分治算法（divide-and-conquer algorithm）的伪代码, 回答下面问题

```latex
DoSomething(A,p,r)

n := r-p+1
```

note：$ceil(2n/3)=\left\lceil {2n/3} \right\rceil$；$:=$ 表示赋值, 等价于 $\to$；A是一个包含n的整数元素的数组, 

1. 写出该算法时间复杂度的递归公式, 并求解其对应的渐进表示
2. 描述一下该算法的功能, 并判断是否是最高效的解决方案

idea:它先处理前 \(2/3\)，再处理后 \(2/3\)，最后再处理一次前 \(2/3\)。这样反复做以后，数组可以排成升序。

````python
# my algorithm time complexity is: Θ(n^(log_{3/2}3))，约 Θ(n^2.71)
from math import ceil
def do_something(A, p=0, r=None):
    # 把 A[p:r+1] 排成升序，p、r 都是 0-based 下标
    if r is None:
        r = len(A) - 1

    n = r - p + 1
    if n <= 1:
        return A

    if n == 2 and A[p] > A[r]:
        A[p], A[r] = A[r], A[p]
    elif n >= 3:
        m = ceil(2 * n / 3)
        do_something(A, p, p + m - 1)
        do_something(A, r - m + 1, r)
        do_something(A, p, p + m - 1)
    return A
arr = [5, 1, 4, 2, 8, 3]
print(do_something(arr.copy()))
````

## 问题 3

给定一个包含n个元素的数组 `profits` , 它的第 `i` 个元素 `profits[i]` 表示一支股票第 `i` 天的**收益**（正数表示涨, 负数表示跌）。你只能选择 **某一天** 买入这只股票, 并选择在 **未来的某一个不同的日子** 卖出该股票。

1. 设计一个算法来计算你所能获取的最大利润和对应买入和卖出的日期。请分析算法方案, 计算其时间复杂度, 并且使用python编程实现该算法。

2. \* 设计一个时间复杂度为 $O(n)$的算法实现该算法

e.g. :
---
profits=[3,2,1,-7,5,2,-1,3,-1], 第5天买入, 第8天卖出, 收益最大：9

idea:这题是在找一段连续天数的最大累计收益。也就是选一个开始日期和结束日期，让 profits[start] + ... + profits[end] 最大。
从左到右扫一遍：维护“以今天结尾的最大收益”。如果前面累计收益已经小于 0，那就无法继续，就从今天重新开始。扫的过程中顺手记录最大值和对应的买入、卖出日期。
时间复杂度：(O(n))；额外空间复杂度：(O(1))。

````python
# my algorithm time complexity is: O(n)，space complexity is: O(1)
def max_profit_days(profits):
    if not profits:
        return 0, None, None
    best_sum = profits[0]
    best_l = best_r = 0
    cur_sum = profits[0]
    cur_l = 0
    for i in range(1, len(profits)):
        # 如果前面那段已经是负的，就从今天重新开始更划算
        if cur_sum < 0:
            cur_sum = profits[i]
            cur_l = i
        else:
            cur_sum += profits[i]

        if cur_sum > best_sum:
            best_sum = cur_sum
            best_l, best_r = cur_l, i
    # 日期按题目习惯用 1-based
    return best_sum, best_l + 1, best_r + 1
profits = [3, 2, 1, -7, 5, 2, -1, 3, -1]
ans, buy_day, sell_day = max_profit_days(profits)
print(f"第{buy_day}天买入，第{sell_day}天卖出，最大收益：{ans}")
````

## 问题 4

给定一个包含不同整数元素的数组 $ A[1..n]$ ,并且满足条件：$A[1]>A[2]$ 并且 $A[n-1]<A[n]$; 规定：如果一个元素比它两边的邻居元素都小, 即：$A[x]<A[x-1], A[x]<A[x+1]$ , 称这个元素A[x]为“局部最小”。通过遍历一次数组, 我们可以很容易在 $O(n)$的时间复杂度下找到一个局部最小值, 


1. 分析该问题, 设计一个算法在$O(logn)$的时间复杂度下找到一个局部最小(返回数值), 要求：分析算法设计思路, 并且使用python编程实现
2. \* 设计算法找出所有局部最小值, 分析算法设计思路, 并使用python编程实现

e.g.:
---
A=[9, 3, 7, 2, 1, 4, 5 ] 时,  局部最小元素为 3, 1

idea:
要找一个局部最小值，可以用二分。看中点 `mid`：
- 如果 `A[mid]` 比左右都小，直接返回；
- 如果左边更小，说明往左走一定能遇到一个局部最小；
- 如果右边更小，说明往右走一定能遇到一个局部最小。
题目给了 `A[1] > A[2]` 和 `A[n-1] < A[n]` 这样的边界条件，所以中间一定存在局部最小。二分每次砍掉一半，时间复杂度是 \(O(\log n)\)，空间复杂度是 \(O(1)\)。
如果要找所有局部最小值，就只能把中间位置都看一遍，因为答案数量本身可能接近 \(n/2\)。所以扫描一次，时间复杂度 \(O(n)\)，空间复杂度 \(O(k)\)，其中 \(k\) 是局部最小值个数。

````python
# my algorithm time complexity is:
#   find_one_local_min: O(log n)
#   find_all_local_min: O(n)

def find_one_local_min(A):
    # 返回任意一个局部最小值。题目保证 A[0] > A[1] 且 A[-2] < A[-1]
    if len(A) < 3:
        return None
    left, right = 1, len(A) - 2
    while left <= right:
        mid = (left + right) // 2
        if A[mid] < A[mid - 1] and A[mid] < A[mid + 1]:
            return A[mid]
        elif A[mid - 1] < A[mid]:
            right = mid - 1
        else:
            left = mid + 1
    return None

def find_all_local_min(A):
    res = []
    for i in range(1, len(A) - 1):
        if A[i] < A[i - 1] and A[i] < A[i + 1]:
            res.append(A[i])
    return res
A = [9, 3, 7, 2, 1, 4, 5]
print("一个局部最小值：", find_one_local_min(A))
print("所有局部最小值：", find_all_local_min(A))
````

给定包含n个不同数字的一组数, 寻找一种基于比较的算法在这组数中找到k个最小的数字, 并按顺序输出它们。

1. 将n个数先进行排序, 然后按顺序输出最小的k个数。要求：选择合适的排序算法实现上述操作, 计算算法时间复杂度, 并使用python编程实现。
2. 建立一个包含这n个数的堆（heap）, 并且调用 k 次Extract-min 按顺序输出最小的k个数。使用往空堆中不断插入元素的方法建立堆, 分析这种方法建堆的时间复杂度, 并使用python编程实现
3. \* 假设数组中包含的数据总数目超过了计算机的存储能力, 请设计一个算法, 找到这堆数据的前k小的数值, 计算时间复杂度, 并使用python实现该算法, 假设计算机一定能存储k个数据。

e.g.：
---
数组arr=[5,4,3,2,6,1,88,33,22,107] 的前3个最小数据为：1, 2, 3

idea:
第一种方法很直接：先把数组排序，再取前 \(k\) 个。这里用归并排序，最坏情况也是 \(O(n\log n)\)，取前 \(k\) 个再花 \(O(k)\)。

第二种方法是从空堆开始把 \(n\) 个数一个个插入小根堆。一次插入是 \(O(\log n)\)，所以建堆是 \(O(n\log n)\)。之后做 \(k\) 次 `extract-min`，每次也是 \(O(\log n)\)，总时间 \(O((n+k)\log n)\)。

第三种是数据太大放不进内存时的做法：只维护一个大小为 \(k\) 的大根堆。堆里始终放目前见过的 \(k\) 个最小值。新数如果比堆顶（当前第 \(k\) 小）还小，就替换堆顶。这样内存只要 \(O(k)\)，总时间约为 \(O(n\log k)\)，最后排序输出再加 \(O(k\log k)\)。

````python
# my algorithm time complexity is:
#   方法1：O(n log n + k)，space O(n)
#   方法2：O((n+k) log n)，space O(n)
#   方法3：O(n log k + k log k)，space O(k)
import heapq
def merge_sort(arr):
    if len(arr) <= 1:
        return arr[:]
    mid = len(arr) // 2
    left = merge_sort(arr[:mid])
    right = merge_sort(arr[mid:])
    i = j = 0
    merged = []
    while i < len(left) and j < len(right):
        if left[i] <= right[j]:
            merged.append(left[i])
            i += 1
        else:
            merged.append(right[j])
            j += 1
    merged.extend(left[i:])
    merged.extend(right[j:])
    return merged

def k_smallest_by_sort(arr, k):
    return merge_sort(arr)[:k]

def k_smallest_by_heap_insert(arr, k):
    heap = []
    for x in arr:
        heapq.heappush(heap, x)     # 从空堆开始插入

    ans = []
    for _ in range(min(k, len(heap))):
        ans.append(heapq.heappop(heap))
    return ans

def k_smallest_stream(data, k):
    if k <= 0:
        return []

    #用负数模拟大根堆
    heap = []
    for x in data:
        if len(heap) < k:
            heapq.heappush(heap, -x)
        elif x < -heap[0]:
            heapq.heapreplace(heap, -x)

    return sorted([-x for x in heap])

arr = [5, 4, 3, 2, 6, 1, 88, 33, 22, 107]
print(k_smallest_by_sort(arr, 3))
print(k_smallest_by_heap_insert(arr, 3))
print(k_smallest_stream(arr, 3))
````

## 问题 6

> 星球碰撞问题：现有n个星球，在同一条直线上运行，如数组A所示，元素的绝对值表示星球的质量，负数表示星球自右向左运动，正数表示星球自左向右运动，当两个星球相撞的时候，质量小的会消失，大的保持不变，**质量相同的两个星球碰撞后自右向左运动的星球消失，自左向右的星球保持不变**，假设所有星球的速度大小相同。
>
> $ A=[23,-8, 9, -3, -7, 9, -23, 22] $

请设计一个算法模拟星球的运行情况，输出最终的星球存续情况（输出一个数组），分析算法设计思路，计算时间复杂度，并基于python编程实现。

e.g.
---
1.  输入： A=[-3,-6,2,8, 5,-8,9,-2,1]， 输出：[-3, -6, 2, 8, 9, 1]

2. 输入：A=[23,-8, 9, -3, -7, 9, -23, 22], 输出：[23, 22]

idea:
用栈来模拟。栈里保存当前还活着的星球。只有一种情况会发生碰撞：栈顶是正数、当前星球是负数，也就是前面的向右走，后面的向左走。

碰撞时比较绝对值：
- 栈顶质量小，栈顶消失，当前星球继续往左和更前面的星球比；
- 栈顶质量大，当前星球消失；
- 质量相同，题目规定负数方向的星球消失，所以当前负数消失，栈顶正数留下。

每个星球最多入栈一次、出栈一次，所以时间复杂度 \(O(n)\)，空间复杂度 \(O(n)\)。

````python
# my algorithm time complexity is: O(n)，space complexity is: O(n)
def asteroid_collision(A):
    stack = []

    for a in A:
        alive = True

        # 只有“栈顶向右、当前向左”才会撞
        while alive and stack and stack[-1] > 0 and a < 0:
            if stack[-1] < -a:
                # 栈顶质量小，栈顶消失，当前星球继续
                stack.pop()
            else:
                # 栈顶质量大或相等时，当前负数星球消失
                alive = False

        if alive:
            stack.append(a)

    return stack

print(asteroid_collision([-3, -6, 2, 8, 5, -8, 9, -2, 1]))
print(asteroid_collision([23, -8, 9, -3, -7, 9, -23, 22]))
````

## 问题 7

> 0-1背包问题：现在有4块大理石，每个大理石对应的重量和价值使用一个元组表示, 即（重量，价值），4块大理石的重量和价值为：[(5,10), (4,40), (6,30), (3, 50)], 假设你有一辆最大承重为9的小推车，请问使用这个小推车装大理石的最大价值为多少

请设计一个算法解决该问题，分析算法设计思路，计算时间复杂度，并基于python编程实现

idea:
这是标准的 0-1 背包。每块大理石只能选或不选，不能切开。设 dp[i][w] 表示只看前 i 块大理石、容量为 w 时能得到的最大价值。

如果第 i 块不装，价值是 dp[i-1][w]；如果装得下，也可以选它，价值是 dp[i-1][w-weight]+value。两者取最大。

时间复杂度：(O(nW))，其中 (W) 是小推车容量；空间复杂度：(O(nW))。这份代码还会顺便把选了哪些大理石找出来。

````python
# my algorithm time complexity is: O(nW)，space complexity is: O(nW)
def knapsack_01(items, capacity):
    # items: [(weight, value), ...]
    # 返回：最大价值、选择的物品下标（从 1 开始）
    n = len(items)
    dp = [[0] * (capacity + 1) for _ in range(n + 1)]

    for i in range(1, n + 1):
        weight, value = items[i - 1]
        for w in range(capacity + 1):
            dp[i][w] = dp[i - 1][w]
            if w >= weight:
                dp[i][w] = max(dp[i][w], dp[i - 1][w - weight] + value)

    # 反推选了哪些
    chosen = []
    w = capacity
    for i in range(n, 0, -1):
        if dp[i][w] != dp[i - 1][w]:
            chosen.append(i)
            w -= items[i - 1][0]

    chosen.reverse()
    return dp[n][capacity], chosen

items = [(5, 10), (4, 40), (6, 30), (3, 50)]
capacity = 9
best_value, chosen = knapsack_01(items, capacity)
print("最大价值：", best_value)
print("选择的大理石编号：", chosen)
````

给你二叉搜索树的根节点 root ，该树中的恰好两个节点的值被错误地交换。请在不改变其结构的情况下，恢复这棵树 。设计算法解决该问题，分析算法设计思路，计算时间复杂度，并基于python编程实现

e.g.:
---
1.

<div style="text-align:center;width:200px;"> <img alt="" src="./fig/hw3q9-1.jpg"> </div>
输入：root = [1,3,null,null,2]， 输出为：[3,1,null,null,2]， 

解释：3 不能是 1 的左孩子，因为 3 > 1 。交换 1 和 3 使二叉搜索树有效。

2.

 <div style="text-align:center;width:200px;"> <img alt="" src="./fig/hw3q9-1.jpg"> </div>
输入： root=[3,1,4,null,null,2]， 输出为：[2,1,4,null,null,3]

解释：2 不能在 3 的右子树中，因为 2 < 3 。交换 2 和 3 使二叉搜索树有效。

idea:
二叉搜索树的中序遍历应该是递增序列。两个节点被交换之后，中序遍历里会出现“前一个值大于后一个值”的逆序位置。

扫描中序序列时：
- 第一次发现 `prev.val > cur.val`，把 `prev` 记成第一个错点；
- 每次发现逆序，都把 `cur` 记成第二个错点；
- 最后交换这两个节点的值。

这样不需要改变树的结构。时间复杂度 \(O(n)\)，迭代中序遍历的栈空间是 \(O(h)\)，其中 \(h\) 是树高。

````python
# my algorithm time complexity is: O(n)，space complexity is: O(h),借助了大模型
from collections import deque
class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right

def build_tree(values):
    """用层序列表建树，None 表示空节点。"""
    if not values:
        return None

    root = TreeNode(values[0])
    q = deque([root])
    i = 1

    while q and i < len(values):
        node = q.popleft()

        if i < len(values) and values[i] is not None:
            node.left = TreeNode(values[i])
            q.append(node.left)
        i += 1

        if i < len(values) and values[i] is not None:
            node.right = TreeNode(values[i])
            q.append(node.right)
        i += 1

    return root

def tree_to_list(root):
    """把树转回层序列表，方便看结果。"""
    if not root:
        return []

    ans = []
    q = deque([root])
    while q:
        node = q.popleft()
        if node is None:
            ans.append(None)
        else:
            ans.append(node.val)
            q.append(node.left)
            q.append(node.right)

    while ans and ans[-1] is None:
        ans.pop()
    return ans

def recover_tree(root):
    first = second = prev = None
    stack = []
    cur = root

    # 迭代中序遍历
    while stack or cur:
        while cur:
            stack.append(cur)
            cur = cur.left

        cur = stack.pop()

        if prev and prev.val > cur.val:
            if first is None:
                first = prev
            second = cur

        prev = cur
        cur = cur.right

    if first and second:
        first.val, second.val = second.val, first.val

    return root

root1 = build_tree([1, 3, None, None, 2])
recover_tree(root1)
print(tree_to_list(root1))

root2 = build_tree([3, 1, 4, None, None, 2])
recover_tree(root2)
print(tree_to_list(root2))
````

## 问题 9 \*
给定 `n` 个非负整数表示每个宽度为 `1` 的柱子的高度图，计算按此排列的柱子，下雨之后能接多少雨水。



<div style="text-align:center;width:800px;"> <img alt="" src="./fig/hw3q6.png"> </div>

请分析算法设计思路，计算时间复杂度，并基于python编程实现

idea:
每个位置能接多少水，取决于它左边最高柱子和右边最高柱子中较矮的那个。为了不额外开两个数组，可以用双指针。

`left` 和 `right` 从两端往中间走，同时维护 `left_max` 和 `right_max`。如果左边最高值较小，就先处理左指针，因为它能接多少水已经由 `left_max` 决定了；右边同理。

时间复杂度 \(O(n)\)，空间复杂度 \(O(1)\)。

````python
# my algorithm time complexity is: O(n)，space complexity is: O(1)
def trap(height):
    if not height:
        return 0
    left, right = 0, len(height) - 1
    left_max = right_max = 0
    water = 0
    while left < right:
        if height[left] <= height[right]:
            left_max = max(left_max, height[left])
            water += left_max - height[left]
            left += 1
        else:
            right_max = max(right_max, height[right])
            water += right_max - height[right]
            right -= 1

    return water
print(trap([0, 1, 0, 2, 1, 0, 1, 3, 2, 1, 2, 1]))
print(trap([4, 2, 0, 3, 2, 5]))
````

## 问题 10

给你一个二叉树的根节点 root ，树中每个节点都存放有一个 0 到 9 之间的数字。每条从根节点到叶节点的路径都代表一个数字：

> 例如，从根节点到叶节点的路径 1 -> 2 -> 3 表示数字 123 。

设计一个算法计算从根节点到叶节点生成的所有数字之和 ，分析算法设计思路，计算时间复杂度，并基于python编程实现

叶节点:是指没有子节点的节点。

e.g.:
---
1.

 <div style="text-align:center;width:200px;"> <img alt="figure" src="./fig/hw3q10-1.jpg"> </div>

 输入：root = [1,2,3], 输出：25

解释：从根到叶子节点路径 1->2 代表数字 12，1->3 代表数字 13，因此和为 12 + 13 = 25 。

2.

 <div style="text-align:center;width:200px;"> <img alt="figure" src="./fig/hw3q10-2.jpg"> </div>

 输入：root = [4,9,0,5,1], 输出：1026

解释：从根到叶子节点路径 4->9->5 代表数字 495，4->9->1 代表数字 491，4->0 代表数字 40。因此和为 495 + 491 + 40 = 1026 。

idea:
从根到叶子的路径可以边走边算。假设走到当前节点之前的数字是 `cur`，当前节点值是 `x`，那么新数字就是 `cur * 10 + x`。

深度优先搜索整棵树：
- 如果到了叶子节点，就把当前路径数字返回；
- 如果不是叶子，就分别算左子树和右子树的和，再加起来。

每个节点访问一次，所以时间复杂度 \(O(n)\)。递归栈深度是树高，空间复杂度 \(O(h)\)。

````python
# my algorithm time complexity is: O(n)，space complexity is: O(h),借助了大模型
from collections import deque
class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right

def build_tree(values):
    if not values:
        return None

    root = TreeNode(values[0])
    q = deque([root])
    i = 1

    while q and i < len(values):
        node = q.popleft()

        if i < len(values) and values[i] is not None:
            node.left = TreeNode(values[i])
            q.append(node.left)
        i += 1

        if i < len(values) and values[i] is not None:
            node.right = TreeNode(values[i])
            q.append(node.right)
        i += 1

    return root

def sum_numbers(root):
    def dfs(node, cur):
        if node is None:
            return 0

        cur = cur * 10 + node.val

        if node.left is None and node.right is None:
            return cur

        return dfs(node.left, cur) + dfs(node.right, cur)

    return dfs(root, 0)

root1 = build_tree([1, 2, 3])
print(sum_numbers(root1))
root2 = build_tree([4, 9, 0, 5, 1])
print(sum_numbers(root2))
````
